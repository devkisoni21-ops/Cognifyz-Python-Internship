# Cognifyz - Task 1: Text-based Game

```python


def quiz_game():
    print("=== Python Quiz Game ===")
    score = 0

    # Rule 1: Questions
    print("\nQ1. Python ka creator kaun hai?")
    print("a) Guido van Rossum  b) Elon Musk")
    ans1 = input("Your answer: ").lower()
    if ans1 == 'a':
        print("Correct!")
        score += 1
    else:
        print("Wrong!")

    print("\nQ2. 2 + 2 * 2 kitna hota hai?")
    ans2 = input("Your answer: ")
    if ans2 == '6':
        print("Correct!")
        score += 1
    else:
        print("Wrong! Answer is 6")

    # Final result using conditional statements
    print(f"\nYour final score is {score}/2")
    if score == 2:
        print("You are a genius!")
    elif score == 1:
        print("Good try!")
    else:
        print("Better luck next time!")

quiz_game()
```

# Cognifyz - Task 2: Number Pattern

```python

n = int(input("Enter number of rows for pyramid: "))


for i in range(1, n + 1):
    
    print(" " * (n - i), end="")
    
    for j in range(1, i + 1):
        print(j, end=" ")
    print()  

print("\nPattern verified successfully!")
```

# Cognifyz - Task 3: CRUD To-Do App - Easy Version     

```python


tasks = []

while True:
    print("\n--- TO-DO LIST ---")
    print("1. Add Task (Create)")
    print("2. Show Tasks (Read)")
    print("3. Update Task (Update)")
    print("4. Delete Task (Delete)")
    print("5. Exit")

    choice = input("Enter choice (1-5): ")

    
    if choice == '1':
        task = input("Enter new task: ")
        tasks.append(task)
        print("Task added!")

    
    elif choice == '2':
        if len(tasks) == 0:
            print("No tasks yet!")
        else:
            print("\nYour Tasks:")
            for i in range(len(tasks)):
                print(f"{i}. {tasks[i]}")

    
    elif choice == '3':
        num = int(input("Which task number to update? "))
        if 0 <= num < len(tasks):
            new_task = input("Enter new task name: ")
            tasks[num] = new_task
            print("Task updated!")
        else:
            print("Invalid number!")

    
    elif choice == '4':
        num = int(input("Which task number to delete? "))
        if 0 <= num < len(tasks):
            tasks.pop(num)
            print("Task deleted!")
        else:
            print("Invalid number!")

    
    elif choice == '5':
        print("Bye! All tasks:", tasks)
        break

    else:
        print("Wrong choice, try again")
```
